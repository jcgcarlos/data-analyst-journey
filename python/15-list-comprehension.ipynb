## List Comprehensions
A different way of building out list using a different notation. A one line <code>for</code> loop build inside of brackets

```python
# Returns every letter in a string
letters = [x for x in 'letters']
```

```python
letters
```

```python
# You can also use operators inside the bracker
square = [x**2 for x in range(0,11)]
```

```python
square
```

```python
# Check for even numbers in a range
# if statement is placed after the for loop
even = [x for x in range(11) if x % 2 == 0] 
```

```python
even
```

```python
# Not limited to simple arithmethic
# Convert Celsius to Fahrenheit
celsius = [0,10,20.1,34.5]

fahrenheit = [((9/5)*temp + 32) for temp in celsius ]

fahrenheit
```

```python
# Nested list comprehension
nestedlc = [ x**2 for x in [x**2 for x in range(11)]]
nestedlc
```
