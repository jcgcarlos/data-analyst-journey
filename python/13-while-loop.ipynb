## While loops

Repeats a block of code **as long as a condition remains True**.
Useful when you don't know in advance how many iterations you need —
unlike a `for` loop where you iterate over a known sequence.

**Syntax:**
```python
while condition:
    # code to repeat
    # must include something that eventually makes condition False
```

**Key controls inside a while loop:**
- `break` — exit the loop immediately
- `continue` — skip the rest of this iteration, go back to the top
- `pass` — do nothing (placeholder)

````python
# --- Core mechanics ---

# Basic while loop: runs until x reaches 10
x = 0
while x < 10:
    print(f'x is currently: {x}')
    x += 1          # without this, loop runs forever (infinite loop!)
````

````python
# else block: runs once when the condition becomes False naturally
x = 0
while x < 5:
    x += 1
else:
    print('Loop finished — condition is now False')
````

````python
# break: exit early when a condition is hit mid-loop
x = 0
while x < 10:
    if x == 3:
        print('Stopping early at x == 3')
        break
    x += 1
````

````python
# continue: skip a specific iteration, keep looping
x = 0
while x < 5:
    x += 1
    if x == 3:
        continue    # skips the print below when x is 3
    print(x)        # prints 1, 2, 4, 5 — 3 is skipped
````

## Practice exercises

**Business problem:**
You're monitoring a support queue at a BPO. Tickets come in one at a time
and each one takes a random number of minutes to resolve. You need to track
how many tickets were closed before cumulative handle time exceeds a shift cap.

````python
# Exercise 1 — Basic while loop
# A queue starts with 50 tickets. Agents close 7 tickets per hour.
# Print the queue count at the start of each hour until it hits zero or below.

queue = 50
hour = 1

while queue > 0:
    print(f'Hour {hour}: {queue} tickets remaining')
    queue -= 7
    hour += 1

print('Queue cleared.')
````

````python
# Exercise 2 — while + break
# Monitor daily AHT (average handle time). Loop through each day's AHT.
# Stop immediately if AHT exceeds 8 minutes (SLA breach).

daily_aht = [5.2, 6.1, 7.4, 8.3, 6.9]
index = 0

while index < len(daily_aht):
    aht = daily_aht[index]
    if aht > 8:
        print(f'SLA breach on day {index + 1}: AHT = {aht} mins. Stopping review.')
        break
    print(f'Day {index + 1}: AHT = {aht} mins — within SLA')
    index += 1
````

````python
# Exercise 3 — while + continue
# Same daily AHT list. Print only the days that are within SLA.
# Skip (but don't stop at) any day that breaches 8 minutes.

index = 0

while index < len(daily_aht):
    aht = daily_aht[index]
    index += 1          # increment before continue to avoid infinite loop
    if aht > 8:
        continue        # skip breach days silently
    print(f'Day {index}: AHT = {aht} mins')
````

## Notes

- The condition must eventually become False — always make sure something
  inside the loop changes the variable being checked, or you get an infinite loop.
- `break` exits the entire loop; `continue` only skips the current iteration.
- When in doubt between `for` and `while`: use `for` when you have a known
  list to iterate; use `while` when you're waiting for a condition to change.
