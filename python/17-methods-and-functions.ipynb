## Functions

A function is a useful device that groups together a set of statements so they can be run more than once. They can also let us specify parameters that can serve as inputs to the functions. On a more fundamental level, functions allow us to not have to repeatedly write the same code again and again.

### When to use functions?

You should use functions when you plan on using a block of code multiple times. The function will allow you to call the same block of code without having to write it multiple times.


### def keyword

```python
def name_of_function(arg1,arg2):
    '''
    This is where the function's Document String (docstring) goes.
    When you call help() on your function it will be printed out.
    '''
    # Do stuff here
    # Return desired result
```

Example

```python
def add_numbers(num1,num2):
    return num1+num2
```

```python
# Calling the function with desired parameters (arguments) inside ()
add_numbers(1,2) 
```

## Adding Logic to Internal Function Operations

### Check if a number is even 

```python
def even_check(num):
    # Return boolean if a number is divisible by 2
    return num%2 == 0
```

```python
even_check(2)
```

```python
even_check(1)
```

### Check if any number in  a list is even

Return a boolean indicating if **any** number in a list is even

Note here how **return** breaks out of the loop and exits the function

```python
def even_check_list(numlist):
    for num in numlist:
        if num%2 == 0:
            return True
        else:
            pass
```

```python
even_check_list([1,2,5])
```

```python
even_check_list([1,3,5])
```

```python
def even_check_list(numlist):
    # Return a True if any number in a list is even and false if there is none
    for num in numlist:
        if num%2 == 0:
            return True
        else:
            pass
    # Notice the placement of return false it is aligned with the for loop
    # The reason for this is that we want the for loop to finish looping in all of the numbers first
    # and return False if there's no even in the list    
    return False
```

```python
even_check_list([1,5,7])
```

```python
even_check_list([2,1,1,1,1])
```

```python
def check_even_list(numlist):
    # Return the even numbers in the list
    # Placeholder list
    even_number_list = []
    for num in numlist:
        if num%2 == 0:
            # Notice that there is no return here because if
            # return is added here then the loop will exit and will
            # only return the first even number
            even_number_list.append(num)
        else:
            pass  

    return even_number_list
```

```python
check_even_list([1,2,6,7])
```

```python
check_even_list([1,7])
```

### Returning Tuples for Unpacking

```python
stock_prices = [('AAPL',200),('GOOG',300),('MSFT',400)]
```

```python
for item in stock_prices:
    print(item)
```

```python
for stock,price in stock_prices:
    print(stock)
```

```python
for stock,price in stock_prices:
    print(price)
```

Functions often also return tuples, to easily return multiple results for later use

```python
work_hours = [('Abby',100),('Billy',400),('Cassie',800)]
```

```python
def employee_check(work_hours):
    
    # Set some max value to intially beat, like zero hours
    current_max = 0
    # Set some empty value before the loop
    employee_of_month = ''
    
    for employee,hours in work_hours:
        if hours > current_max:
            current_max = hours
            employee_of_month = employee
        else:
            pass
    
    # Notice the indentation here
    return (employee_of_month,current_max)
```

```python
employee_check(work_hours)
```
