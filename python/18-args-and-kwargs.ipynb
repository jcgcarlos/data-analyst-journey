## *args and **kwargs

### *args — Variable Positional Arguments
Lets a function accept any number of positional arguments (no labels).
Python collects them into a **tuple** inside the function.

Use it when you don't know how many values will be passed, but they're all the same *kind* of thing (amounts, scores, items).

---

```python
def total(*args):
    return sum(args)

total(100, 200, 300)
```

### **kwargs — Variable Keyword Arguments
Lets a function accept any number of keyword arguments (`name=value` pairs).
Python collects them into a **dictionary** inside the function.

Use it when you want optional, named metadata — things that vary per call.

---

```python
def show_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_info(currency="PHP", region="NCR")
# → currency: PHP
# → region: NCR
```

### Using Both Together
Order matters: `regular args → *args → **kwargs`

```python
def summary(label, *args, **kwargs):
    print(f"Label: {label}")
    print(f"Values: {args}")
    print(f"Metadata: {kwargs}")
```

```python
summary('Dogs', 1, 2, 3, breed = 'Shi Tzu')
```

Exercise

Scenario: You're building a simple internal tool for a fintech company. Analysts need to generate quick summary reports for any transaction batch.

Write a function called transaction_summary that:

Takes a batch_name as a regular required argument
Uses *args to accept any number of transaction amounts (as numbers)
Uses **kwargs to accept optional metadata like currency, region, or analyst

Prints:
1. The batch name
2. Total number of transactions
3. Total amount
4. Any metadata that was passed in

```python
def transaction_summary(batch_name, *args, **kwargs):
    print(f"Batch: {batch_name}")
    print(f"Transactions: {len(args)}")
    print(f"Total Amount: {sum(args):,.2f}")

    for key,value in kwargs.items():
        print(f"    {key} = {value}")
```

```python
transaction_summary(
    "Batch_Oct01",
    500, 1200, 340, 875,
    currency="PHP",
    region="NCR",
    analyst="JC"
)
```
