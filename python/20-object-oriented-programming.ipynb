## Classes, Objects, Attributes, and Methods

### The Big Picture — What OOP Actually Is

OOP (Object-Oriented Programming) is a way of organizing code around **things**
instead of a long list of instructions. Each "thing" (object) carries its own
data and its own actions.

As a data analyst, you won't build classes from scratch every day — but you use
them constantly without realizing it. Every time you write `df.groupby()` or
`df.describe()`, you're calling a method on a DataFrame object. Someone wrote a
`DataFrame` class so you could use it without thinking about it.

Understanding OOP means you can write cleaner scripts, read library docs properly,
and eventually build reusable tools for your own workflow.

---

### Class — The Blueprint

A class is a template. It defines what an object looks like and what it can do.
You define it once, then create as many objects from it as you need.

**Syntax:**
```python
class ClassName:
    def __init__(self, param1, param2):
        self.param1 = param1
        self.param2 = param2
```

`__init__` is the constructor — it runs automatically when you create a new object.
`self` refers to the object being created. Always the first parameter in any method.

---

### Object — The Actual Thing

An object is one instance built from a class. Same blueprint, different data.

```python
class Transaction:
    def __init__(self, transaction_id, amount, status):
        self.transaction_id = transaction_id
        self.amount = amount
        self.status = status

# Creating objects from the class
txn_001 = Transaction("TXN-001", 250.00, "success")
txn_002 = Transaction("TXN-002", 180.00, "failed")
```

`txn_001` and `txn_002` are two separate objects — same blueprint, different values.

---

### Attributes — The Data an Object Holds

Attributes are the variables attached to an object. You access them with dot notation.

```python
print(txn_001.transaction_id)  # → TXN-001
print(txn_001.amount)          # → 250.0
print(txn_001.status)          # → success
```

**Real-world parallel:**
Think of a bank transaction form. The class is the blank form.
Each filled-in form is an object. The fields (amount, status, merchant) are attributes.

---

### Methods — What an Object Can Do

Methods are functions defined inside a class. They operate on the object's own data.

```python
class Transaction:
    def __init__(self, transaction_id, amount, status):
        self.transaction_id = transaction_id
        self.amount = amount
        self.status = status

    def summary(self):
        print(f"{self.transaction_id} | ₱{self.amount:.2f} | {self.status}")

    def is_failed(self):
        return self.status == "failed"


txn_001 = Transaction("TXN-001", 250.00, "success")
txn_002 = Transaction("TXN-002", 180.00, "failed")

txn_001.summary()       # → TXN-001 | ₱250.00 | success
txn_002.summary()       # → TXN-002 | ₱180.00 | failed

print(txn_002.is_failed())  # → True
```

Methods make objects self-contained. Instead of writing separate functions for
every transaction, you define the logic once inside the class and reuse it.

---

### Why This Matters for Data Analysis

| Concept | Where you already use it |
|---|---|
| Class | `pd.DataFrame`, `sklearn.LinearRegression` |
| Object | `df = pd.read_csv("file.csv")` — df is an object |
| Attributes | `df.shape`, `df.columns`, `df.dtypes` |
| Methods | `df.groupby()`, `df.fillna()`, `df.describe()` |

When you build your own classes, the use cases are:
- Wrapping reusable report logic (e.g., a `MerchantReport` class)
- Representing business entities cleanly (User, Transaction, Campaign)
- Building automation scripts that are easier to maintain and extend

Exercise: 

Build a UserAccount class

You're an analyst at a fintech app. You need a simple way to represent a user account and track their balance.
Build a class called UserAccount that:

Takes in: user_id, name, balance
Has a method deposit(amount) that adds to the balance
Has a method summary() that prints something like:

U-001 | Juan dela Cruz | Balance: ₱1,500.00

````python
class UserAccount:
    def __init__ (self, user_id, name, balance):
        self.user_id = user_id
        self.name = name
        self.balance = balance

    def deposit (self, deposit_amount):
        self.balance += deposit_amount

    def withdraw (self, withdrawal_amount):
        if self.balance >= withdrawal_amount:
            self.balance -= withdrawal_amount
        else:
            print(f"Insufficient funds. Current balance: {self.balance:.2f}")

    def summary (self):
        print(f"{self.user_id} | {self.name} | Balance: {self.balance:.2f}")
````

````python
user_1 = UserAccount("U-001", "Juan dela Cruz", 500)
user_1.withdraw(2000)
user_1.summary()
````

````python
user_1 = UserAccount("U-001", "Juan dela Cruz", 500)
user_1.deposit(2000)
user_1.summary()
````
