Exercise 2 — if/elif/else chain

Classify a ticket by priority level:

```python
resolution_time = 5  # in hours

# If under 2 hours: "Priority 1 – Critical"
# If under 8 hours: "Priority 2 – High"
# If under 24 hours: "Priority 3 – Medium"
# Otherwise: "Priority 4 – Low"

if resolution_time < 2:
    print('Priority 1 - Critical')
elif resolution_time < 8:
    print('Priority 2 - High')
elif resolution_time < 24:
    print('Priority 3 - Medium')
else:
    print('Priority 4 - Low')
```

Exercise 4 — for loop + if/else combined

Loop through the same list and flag each day:

```python
daily_calls = [312, 289, 401, 375, 290]
target = 300

# For each day's volume:
#   If it meets or exceeds 300, print: "Day X: 312 – On target"
#   Otherwise print: "Day X: 289 – Below target"
# Use the index to show "Day 1", "Day 2", etc.

for index, calls in enumerate(daily_calls, start=1):
    if calls >= target:
        print(f'Day {index}: {calls} - On target')
    else:
        print(f'Day {index}: {calls} - Below target')
```

Exercise 5 — running total with a loop

Calculate the total calls for the week using a loop (not sum()):

```python
daily_calls = [312, 289, 401, 375, 290]
total = 0

# Loop through the list and accumulate the total
# Print the final total at the end

for calls in daily_calls:
    total = total + calls
print(total)
```
