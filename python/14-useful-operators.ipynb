## Useful Operators

## range 
-function that allows you to *generate* a list of integers

syntax: range(start,stop,step)

**Note:**
this is a generator so to actually have a list you need to cast it to a list with **list()**

```python
range(0,12)
```

```python
# start at 0 upto 12 but not including 12
list(range(0,12))
```

```python
# third parameter is the step size
list(range(0,12,2))
```

## enumerate

used to keep track of how many loops you've gone through

```python
for i,num in enumerate(range(0,6)):
    print(f"At index {i} the number is {num}")
```

## zip

to quickly create a list of tuples you can use *zip* to combine together two list

```python
num = [1,2,3,4,5,6,7]
letter = ['a', 'b', 'c', 'd', 'e']
```

```python
# same with range you need to cast it to a list using list()
zip(num,letter)

list(zip(num,letter))
```

## in
used to check if an object is in a list

```python
'x' in ['x', 'y', 'z']
```

```python
0 in [1,2,3]
```

## not in
combing the not operator to check if some object or variable is not present in a list

```python
'x' not in ['x', 'y', 'z']
```

```python
0 not in [1,2,3]
```

# min and max
check the minimum and maximum in a list

```python
mylist = [1,10,100,1000,10000]
```

```python
min(mylist)
```

```python
max(mylist)
```

## random
built in library with useful functions

```python
from random import shuffle
```

```python
# shuffles the list "inplace" meaning it won't return
# anything until the list gets passed
shuffle(mylist)
```

```python
from random import randint
```

```python
# Return a random integer in range [a,b], including both end points
randint(0,1000)
```
