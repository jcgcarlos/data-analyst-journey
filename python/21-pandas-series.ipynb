## Pandas Series

### Series — A Single Column with a Label

A Series is a one-dimensional labeled array. Think of it as a single column
from a spreadsheet — it has values, and each value has a label (the index).

As an analyst, you'll rarely build one from scratch. But every column you pull
from a DataFrame is a Series — so understanding it changes how you read and
write pandas code.

**Syntax:**
pd.Series(data, index=[labels])

**Basic example — fintech:**
You have account balances for three users.

```python
import pandas as pd

balances = pd.Series([1500, 3200, 800], index=["Juan", "Maria", "Pedro"])
print(balances)
```

You can pull a value by its label instead of its position.

```python
balances["Juan"]
```

```python
balances["Maria"]
```

### The Index Is What Makes It Useful

A plain Python list only has positions (0, 1, 2...).
A Series has labels — and those labels can be anything meaningful:
a date, a user ID, a merchant name.

That's what makes slicing and aggregation fast and readable.

**Real-world example — fintech:**
You have daily transaction revenue for the first week of January.
You need to pull a specific day and get basic stats.

```python
daily_revenue = pd.Series(
    [45000, 62000, 38000, 71000, 55000],
    index=["2024-01-01", "2024-01-02", "2024-01-03", "2024-01-04", "2024-01-05"]
)
```

```python
daily_revenue["2024-01-02"]
```

```python
daily_revenue.max() 
```

```python
daily_revenue.mean()  
```

```python
daily_revenue.sum() 
```

No loops. No manual math. Just call the method.

This is the kind of thing you'd use when tracking daily active users,
weekly revenue trends, or monthly churn rates.

### Series vs. Plain Python List

| | Python List | Pandas Series |
|---|---|---|
| Has labels | No — only positions | Yes — custom index |
| Built-in math | No | Yes — .sum(), .mean(), .max() |
| Part of DataFrame | No | Yes — every column is a Series |
| Pull by label | No | Yes — s["label"] |

### How This Shows Up in Real Work

Every time you pull a column from a DataFrame, you get a Series back.
That means all the methods above — .sum(), .mean(), .max() — are available
immediately on any column you work with.

**Example — pulling a column from a DataFrame:**

```python
# df["transaction_amount"] returns a Series
# From there, you can immediately aggregate:

df["transaction_amount"].sum()   # total revenue
df["transaction_amount"].mean()  # average transaction value
df["transaction_amount"].max()   # highest single transaction
```
