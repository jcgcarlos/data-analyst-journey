## Lambda, Map, and Filter

### Lambda — The Throwaway Function

A lambda is an anonymous (unnamed) function. Use it when the logic is simple
and you only need it once. No need to write a full `def` block.

**Syntax:**
lambda argument(s): expression

**Regular function vs. lambda:**

```python
# Regular function
def double(x):
    return x * 2

# Same thing as a lambda
double = lambda x: x * 2

double(5)  # → 10
```

Lambdas are most useful when passed directly into another function
like map() or filter() — you'll see this below.

### map() — Transform Every Item

map() applies a function to every item in a list (or any iterable)
and returns a new iterable with the results.

**Syntax:**
map(function, iterable)

**Real-world example — fintech:**
You have a list of transaction amounts in USD.
You need to convert them all to PHP.

```python
usd_amounts = [10, 25, 50, 100, 200]
exchange_rate = 56.5

php_amounts = list(map(lambda x: x * exchange_rate, usd_amounts))
print(php_amounts)
# → [565.0, 1412.5, 2825.0, 5650.0, 11300.0]
```

Without lambda, you'd have to define a named function first.
With lambda, it's one clean line.

### filter() — Keep Only What Passes the Test

filter() applies a function that returns True or False to each item,
and keeps only the items where the result is True.

**Syntax:**
filter(function, iterable)

**Real-world example — e-commerce:**
You have a list of order values and you want to flag high-value orders
(above ₱5,000) for priority fulfillment.

```python
order_values = [1200, 4800, 6500, 3200, 9100, 500]

high_value_orders = list(filter(lambda x: x > 5000, order_values))
print(high_value_orders)
# → [6500, 9100]
```

### When Do You Actually Use These?

| Tool     | Use it when...                                              |
|----------|-------------------------------------------------------------|
| lambda   | The function is simple and you don't need to reuse it       |
| map()    | You want to transform every item in a collection            |
| filter() | You want to keep only items that meet a condition           |

Exercise 1

You have a list of transaction amounts in PHP. Apply a 2.5% processing fee to each one and return the updated amounts.

```python
transactions = [500, 1200, 3400, 800, 15000]

# Use map() + lambda to apply a 2.5% fee to each transaction
# Expected output: [512.5, 1230.0, 3485.0, 820.0, 15375.0]
```

Exercise 2

A GCash risk team flags transactions above ₱10,000 as "high-value" for review. 

From the list below, filter out only the high-value ones — then apply a 1% compliance hold on each flagged amount.

```python
transactions = [500, 12000, 3400, 45000, 800, 11500, 250]

# Step 1: Use filter() + lambda to get transactions above 10,000
# Step 2: Use map() + lambda to apply a 1% hold on each flagged amount
# Expected output: [12120.0, 45450.0, 11615.0]
```

```python
high_value = list(filter(lambda x: x > 10000, transactions))
results = list(map(lambda x: x * 1.01, high_value))
print(results)
```
